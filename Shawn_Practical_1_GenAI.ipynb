Natural Language Processing (NLP) enables computers to process and understand human language. This practical applies the common stages of an NLP pipeline to a sample paragraph.

```python
import nltk
import spacy
from nltk.tokenize import sent_tokenize, word_tokenize
from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer

for package in ['punkt', 'punkt_tab', 'stopwords', 'wordnet', 'omw-1.4', 'averaged_perceptron_tagger', 'averaged_perceptron_tagger_eng']:
    nltk.download(package, quiet=True)

nlp = spacy.load('en_core_web_sm')
print('All libraries and language resources are ready.')
```

```python
text = "Apple CEO Tim Cook visited New Delhi on Monday. He discussed new technology and educational projects in India."
print('Input Text:')
print(text)
```

```python
sentences = sent_tokenize(text)
tokens = word_tokenize(text)

print('Sentence Tokenization:', sentences)
print('\nWord Tokenization:', tokens)
```

```python
stop_words = set(stopwords.words('english'))
filtered_words = [word for word in tokens if word.isalpha() and word.lower() not in stop_words]

stemmer = PorterStemmer()
stemmed_words = [stemmer.stem(word) for word in filtered_words]

lemmatizer = WordNetLemmatizer()
lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print('After Stopword Removal:', filtered_words)
print('\nStemmed Words:', stemmed_words)
print('\nLemmatized Words:', lemmatized_words)
```

```python
pos_tags = nltk.pos_tag(tokens)
print('POS Tags:')
print(pos_tags)

doc = nlp(text)
entities = [(entity.text, entity.label_) for entity in doc.ents]
print('\nNamed Entities:')
print(entities)

grammar = r'NP: {<DT>?<JJ.*>*<NN.*>+}'
chunk_parser = nltk.RegexpParser(grammar)
chunk_tree = chunk_parser.parse(pos_tags)
print('\nChunking Result:')
print(chunk_tree)
```

## Conclusion
Thus, an NLP pipeline was successfully implemented using NLTK and spaCy. The program performed sentence and word tokenization, stopword removal, stemming, lemmatization, POS tagging, named entity recognition, and noun-phrase chunking.
